```python
# create class bankaccount
class Bankaccount:
  #class attribute
  Branch = 'Pune'
  IFSC_CODE = 'PUNE001'
  counter = 0

  #instance attribute
  def __init__(self,name,acc_number,balance):
    self.name = name
    self.acc_number = acc_number
    self.balance = balance
    Bankaccount.counter += 1

  #instance methods
  def deposit(self,amount):
    self.balance += amount
    return f'{amount} credited in your bank account successfully. current balance is {self.balance}'

  def withdraw(self,amount):
    if self.balance >= amount:
      self.balance -= amount
      return f'{amount} debited from your bank account successfully. current balance is {self.balance}'
    else:
      return f'insufficient balance,The current balance is {self.balance}'

  def get_balance(self):
    return f'current balance is {self.balance}'

  @classmethod
  def get_number_of_accounts(cls):
    return cls.counter
```

##Method overriding

Till now we have inherit the method from super class to base class . but that was exactly the same method . What if the method name is same but the output of both the method is differant then?

```python
# for example as follows
class Employee:
    def __init__(self, name, base_pay):
        self.name = name
        self.base_pay = base_pay
# define a method to return the salary of employee
    def get_pay(self):
        return self.base_pay
```

```python
employee3 = Employee('vaibhav',5000)
```

```python
employee3.get_pay()
```

```python
class SalesEmployee(Employee):
    def __init__(self, name, base_pay, sales_incentive):
        self.name = name
        self.base_pay = base_pay
        self.sales_incentive = sales_incentive
```

```python
# lets create an instance of the salesEmployee class and get the salary which is the addition of base pay and sales_incentive
sales_employee1 = SalesEmployee('vaibhav',5000,1000)
```

```python
sales_employee1.get_pay()
```

The get_pay() method returns only the base_pay, not the sum of the base_pay and sales_incentive.

When you call the get_pay() from the instance of the SalesEmployee class, Python executes the get_pay() method of the Employee class, which returns the base_pay.

To include the sales incentive in the pay, you need to redefine the get_pay() method in the SalesEmployee class as follows:

```python
class SalesEmployee(Employee):
    def __init__(self, name, base_pay, sales_incentive):
        self.name = name
        self.base_pay = base_pay
        self.sales_incentive = sales_incentive

    def get_pay(self):
      return self.base_pay + self.sales_incentive
```

```python
employee4 = SalesEmployee('dhanesh',45000,3500)
```

```python
employee4.get_pay()            #calls the get pay method of salesempoloyee class
```

```python
employee3.get_pay()              #calls the get pay method of the employee class
```

What happened here?


when you define the method in the base class which is having the same name as the method of super class, and If you create an instance of the base class class, Python will call the method of the base class, not the super class

we have created get_pay() method in the salesemployee class which is having the same name as the method of super class. because of that when we create an instance of the salesEmployee class, Python will call the get_pay() method of the salesEmployee class, not the get_pay() method of the Employee class.

and when we create the instance of employee it will call the get_pay() method of empoloyee class not the salesemployee class

##Python super

Python super() method is used to inherit common attribute and method from the super class or parent class to child class .

while writing the child class we use this super method to inherit the common attributes and method of the parent class

Lets understand with the example as follows

```python
class Customer:
    def __init__(self, name, base_pay, bonus):
        self.name = name
        self.base_pay = base_pay
        self.bonus = bonus

    def get_pay(self):
        return self.base_pay + self.bonus
```

The Customer class has three instance variables name, base_pay, and bonus. It also has the get_pay() method that returns the total of base_pay and bonus.

Second, define the SalesCustomer class that inherits from the Customer class:

```python
# Not recommended
class SalesCustomer(Customer):
    def __init__(self, name, base_pay, bonus, sales_incentive):
        self.name = name
        self.base_pay = base_pay
        self.bonus = bonus
        self.sales_incentive = sales_incentive

    def get_pay(self):
        return self.base_pay + self.bonus + self.sales_incentive
```

The SalesCustomer class has four instance variables name, base_pay, bonus, and sales_incentive. It has the get_pay() method that overrides the get_pay() method in the Customer class.

IMP OBSERVATION

1.Salescustomer class has some attributes as same as Customer class. SO to avoid dublication we use super method to access the common attribute from the customer class as follows

2.To reference the Customer class in the SalesCustomer class, you use the super(). The super() returns a reference of the parent class from a child class.

```python
# Use super method in the salescustomer class to get the common attribute between customer and salescustomer class
class SalesCustomer(Customer):
  def __init__(self,name,base_pay,bonus,sales_incentive):
    #inherit common attributes from the customer class
    super().__init__(name,base_pay,bonus)
    self.sales_incentive = sales_incentive

  def get_pay(self):
    return self.base_pay + self.bonus + self.sales_incentive
```

here customer1 calls the get_pay() method of the customer class and salesCustomer calls the get_pay() methof of the salescustomer class. this is nothing but the method overridding

If i want to return the result of get_pay() method of customer class for the salescustomer class then we need to call the super method in in the salescustomer class as follows

```python
# Use super method in the salescustomer class to get the common attribute between customer and salescustomer class
class SalesCustomer1(Customer):
  def __init__(self,name,base_pay,bonus,sales_incentive):
    #inherit common attributes from the customer class
    super().__init__(name,base_pay,bonus)
    self.sales_incentive = sales_incentive

  def get_pay(self):
    salary_with_bonus = super().get_pay()
    salary_with_bonusandincentive =  self.base_pay + self.bonus + self.sales_incentive
    return salary_with_bonus,salary_with_bonusandincentive
```

```python
# create parent class callede manager
class Manager:
  def __init__(self,name,department):
    self.name = name
    self.department = department

  def get_manager_details(self):
    return f'Name : {self.name}, Department : {self.department}'

#create another parent class called worker
class Worker:
  def __init__(self, name,wage):
    self.name = name
    self.wage = wage

  def get_wage(self):
    return f' the worker has monthly wages of {self.wage}'


class Contractor(Manager, Worker):
  def __init__(self, name, department, wage, contract_duration):
  # As the contractor class has order of manager and worker . so when we call super() method it will inherit the manager attributes only
  # super().__init__(name, department)

  # and for the worker class we need to call it manually as follows
    # Worker.__init__(self,name, wage)

  # Instead we call the attribute of both the classes manually
    Manager.__init__(self, name, department)
    Worker.__init__(self,name, wage)
    self.contract_duration = contract_duration

  def get_contract_details(self):
    manager_details = self.get_manager_details()
    return f"{manager_details}, Wage: {self.get_wage()}, Contract Duration: {self.contract_duration} months"

  def get_total_earnings(self):
     return self.wage * self.contract_duration


# Example usage
contractor = Contractor("Alice Johnson", "IT", 5000, 12)

print(contractor.get_contract_details())
print("Total Earnings:", contractor.get_total_earnings())
```

```python
class Location:
    def __init__(self, city, country):
        self.city = city
        self.country = country

    def get_location(self):
        return f"Location: {self.city}, {self.country}"


class Weather:
    def __init__(self, temperature, humidity):
        self.temperature = temperature
        self.humidity = humidity

    def get_weather(self):
        return f"Weather: {self.temperature}C, Humidity: {self.humidity}%"


class Environment(Location, Weather):
    def __init__(self, city, country, temperature, humidity, pollution_level):
        # Initialize Location part
        Location.__init__(self, city, country)
        # Initialize Weather part
        Weather.__init__(self, temperature, humidity)
        self.pollution_level = pollution_level

    def get_environment_details(self):
        location_info = self.get_location()
        weather_info = self.get_weather()
        return f"{location_info}, {weather_info}, Pollution Level: {self.pollution_level}"


# Example usage
environment = Environment("New York", "USA", 25, 60, "Moderate")

print(environment.get_environment_details())
```

##Multilevel Inheritance

Multilevel inheritance in Python occurs when a derived class (child class) is created from another derived class, making a chain of inheritance. This means a class inherits from a class which itself inherits from another class, forming a hierarchy of classes.

example

class A:

  pass

class B(A):

  pass

```python
# Base class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def drive(self):
        return f"{self.brand} vehicle is being driven."


# Intermediate class inheriting from Vehicle
class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def drive(self):
        return f"A {self.brand} {self.model} car is being driven."


# Derived class inheriting from Car
class ElectricCar(Car):
    def __init__(self, brand, model, range_per_charge):
        super().__init__(brand, model)
        self.range_per_charge = range_per_charge

    def drive(self):
        return f"A {self.brand} {self.model} electric car is being driven. Range per charge: {self.range_per_charge} miles."


# Example usage
car = ElectricCar("Tesla", "Model S", 350)
print(car.drive())  # Outputs: A Tesla Model S electric car is being driven. Range per charge: 350 miles.
```

Key Points:
Inheritance Chain: ElectricCar inherits from Car, which in turn inherits from Vehicle, creating a multilevel inheritance hierarchy.

Method Overriding: Each subclass (Car and ElectricCar) provides its own implementation of the drive method, allowing for specialization while retaining polymorphic capabilities.

Initialization: Each subclass uses super() to delegate initialization to its immediate parent class (super().__init__(...)), ensuring attributes are set up correctly through the inheritance chain.

```python
class Person:
    def __init__(self, name, age, address):
        self.name = name
        self.age = age
        self.address = address

    def get_info(self):
        return f"Name: {self.name}, Age: {self.age}, Address: {self.address}"

    def greet(self):
        return f"Hello, my name is {self.name}."


class Student(Person):
    def __init__(self, name, age, address, student_id, major):
        super().__init__(name, age, address)
        self.student_id = student_id
        self.major = major

    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Student ID: {self.student_id}, Major: {self.major}"

    def study(self):
        return f"{self.name} is studying {self.major}."


class GraduateStudent(Student):
    def __init__(self, name, age, address, student_id, major, thesis_title, advisor):
        super().__init__(name, age, address, student_id, major)
        self.thesis_title = thesis_title
        self.advisor = advisor

    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Thesis: {self.thesis_title}, Advisor: {self.advisor}"

    def research(self):
        return f"{self.name} is researching on their thesis titled '{self.thesis_title}' under the guidance of {self.advisor}."


# Example usage
person = Person("Alice", 30, "123 Main St")
student = Student("Bob", 20, "456 College Ave", "S12345", "Computer Science")
grad_student = GraduateStudent("Charlie", 25, "789 University Rd", "G67890", "Physics", "Quantum Mechanics", "Dr. Smith")

print(person.get_info())  # Outputs: Name: Alice, Age: 30, Address: 123 Main St
print(person.greet())     # Outputs: Hello, my name is Alice.

print(student.get_info())  # Outputs: Name: Bob, Age: 20, Address: 456 College Ave, Student ID: S12345, Major: Computer Science
print(student.study())     # Outputs: Bob is studying Computer Science.

print(grad_student.get_info())  # Outputs: Name: Charlie, Age: 25, Address: 789 University Rd, Student ID: G67890, Major: Physics, Thesis: Quantum Mechanics, Advisor: Dr. Smith
print(grad_student.research())  # Outputs: Charlie is researching on their thesis titled 'Quantum Mechanics' under the guidance of Dr. Smith.
print(grad_student.greet())
```

##Polymorphism

Poly means many and morph means form. so polymorphism means many forms

This means you can use a single function or method to operate on different kinds of objects in a way that is seamless and interchangeable.

####Key Points of Polymorphism

####1.Same Function Name, Different Behavior(operator overloading):

Polymorphism allows the same function name to be used for different types of objects. For example, a function add(a, b) might add two numbers if they are integers, or concatenate two strings if they are strings.



####2.Method Overriding:

In object-oriented programming, polymorphism often involves method overriding. This is where a subclass provides a specific implementation of a method that is already defined in its superclass. The overridden method in the subclass will be called instead of the one in the superclass.



```python
# example of operator overloading
'''operator overloading means the same operator has differant meaning
here we can see that the + operator behaves differant in both the cases.for the first case it adds the two number
in the second case it concatenate the two strings'''

def add(a,b):
  return a+b

print(add(3,4))
print(add('hello','world'))
```

```python
#Exampl,e of method overridding
# Super class Animal
class Animal:
    def __init__(self, name, age, animal_type):
        self.name = name
        self.age = age
        self.animal_type = animal_type

    def eat(self):
        return f'{self.name} is eating'

    def make_sount(self):
      return ' the animal is making sound'

# Child class Dog inherits from Animal
class Dog(Animal):
    def __init__(self, name, age, animal_type, breed, color, food_type):
        super().__init__(name, age, animal_type)
        self.breed = breed
        self.color = color
        self.food_type = food_type

    def eat(self):
        return f'{self.name} is eating {self.food_type}'

    def make_sound(self):
        return f'{self.name} is barking'

# Child class Goat inherits from Animal
class Goat(Animal):
    def __init__(self, name, age, animal_type, breed, food_type):
        super().__init__(name, age, animal_type)
        self.breed = breed
        self.food_type = food_type

    def eat(self):
        return f'{self.name} is eating {self.food_type}'

    def make_sound(self):
        return f'{self.name} is bleating'

# Example usage
dog = Dog('Buddy', 3, 'Dog', 'Golden Retriever', 'Golden', 'kibble')
goat = Goat('Billy', 2, 'Goat', 'Mountain Goat', 'hay')

print(dog.eat())  # Output: Buddy is eating kibble
print(dog.make_sound())  # Output: Buddy is barking
print(goat.eat())  # Output: Billy is eating hay
print(goat.make_sound())  # Output: Billy is bleating
```

The eat and make sound method is there in the super class still the instances of the dog and goat class uses their respective eat and make sound method to print the output this is called the method overriding and this is because of polymorphism . That is name of the methods are same but has differant working for particular class
